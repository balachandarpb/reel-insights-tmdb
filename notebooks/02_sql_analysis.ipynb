## Question 1: Find all Action movies released after 2015s

```python
query = """
SELECT DISTINCT
    m.title,
    m.release_date,
    g.genre_name

FROM movies m

JOIN movie_genres mg
    ON m.movie_id = mg.movie_id

JOIN genres g
    ON mg.genre_id = g.genre_id

WHERE g.genre_name = 'Action'
AND m.release_date >= '2016-01-01'

ORDER BY m.release_date DESC
"""

cursor.execute(query)

result = cursor.fetchall()

for row in result:
    print(row)
```

## Question 2: Find the top 10 highest-grossing movies, including their genres.

```python
query = """
SELECT
    m.title,
    m.revenue,
    g.genre_name

FROM movies m

JOIN movie_genres mg
    ON m.movie_id = mg.movie_id

JOIN genres g
    ON mg.genre_id = g.genre_id

WHERE m.revenue > 0

ORDER BY m.revenue DESC

LIMIT 10
"""

cursor.execute(query)

result = cursor.fetchall()

for row in result:
    print(row)
```

```python
query = """
SELECT
    m.title,
    m.revenue,
    GROUP_CONCAT(
        DISTINCT g.genre_name
        ORDER BY g.genre_name
        SEPARATOR ', '
    ) AS genres

FROM movies m

JOIN movie_genres mg
    ON m.movie_id = mg.movie_id

JOIN genres g
    ON mg.genre_id = g.genre_id

WHERE m.revenue > 0

GROUP BY m.movie_id, m.title, m.revenue

ORDER BY m.revenue DESC

LIMIT 10
"""

cursor.execute(query)
result = cursor.fetchall()

for row in result:
    print(row)
```

## Question 3 – Calculate the average budget and revenue for each genre.

```python
query = """
SELECT
    g.genre_name,
    AVG(m.budget) AS average_budget,
    AVG(m.revenue) AS average_revenue

FROM movies m

JOIN movie_genres mg
    ON m.movie_id = mg.movie_id

JOIN genres g
    ON mg.genre_id = g.genre_id

WHERE m.budget > 0
AND m.revenue > 0

GROUP BY g.genre_name

ORDER BY average_revenue DESC
"""

cursor.execute(query)

result = cursor.fetchall()

for row in result:
    print(row)

# columns = [
#     "Genre",
#     "Average Budget",
#     "Average Revenue"
# ]

# df = pd.DataFrame(result, columns=columns)

# display(df.round(2))
```

## List the top 10 actors who have appeared in the most movies in the dataset.(cast)

```python
query = """
SELECT
    actor_name,
    COUNT(DISTINCT movie_id) AS total_movies

FROM `cast`

GROUP BY actor_name

ORDER BY total_movies DESC

LIMIT 10
"""

cursor.execute(query)

result = cursor.fetchall()

for row in result:
    print(row)
```

## Question 5 – Find all directors who have directed more than 3 movies, along with their movie count.

```python
query = """
SELECT
    person_name AS director_name,
    COUNT(DISTINCT movie_id) AS total_movies

FROM crew

WHERE job = 'Director'

GROUP BY person_id, person_name

HAVING COUNT(DISTINCT movie_id) > 3

ORDER BY total_movies DESC
"""

cursor.execute(query)

result = cursor.fetchall()

for row in result:
    print(row)
```

## Question 6 List the top 10 most frequently used keywords across all movies.

```python
query = """
SELECT
    keyword_name,
    COUNT(DISTINCT movie_id) AS total_movies

FROM movie_keywords

GROUP BY keyword_id, keyword_name

ORDER BY total_movies DESC

LIMIT 10
"""

cursor.execute(query)

result = cursor.fetchall()

for row in result:
    print(row)
```

## Question 7 – Find movies with a budget above the average budget but revenue below the average revenue.

```python
query = """
SELECT
    title,
    budget,
    revenue

FROM movies

WHERE budget > (
    SELECT AVG(budget)
    FROM movies
    WHERE budget > 0
)

AND revenue < (
    SELECT AVG(revenue)
    FROM movies
    WHERE revenue > 0
)

AND budget > 0
AND revenue > 0

ORDER BY budget DESC
"""

cursor.execute(query)

result = cursor.fetchall()

for row in result:
    print(row)
```

## Question 8 – Find all movies featuring actors who worked in a movie by a specific director.

```python
#Step 1: Find all movies directed by Christopher Nolan

query = """
SELECT DISTINCT
    movie_id,
    person_name,
    job

FROM crew

WHERE person_name = 'Christopher Nolan'
AND job = 'Director'
"""

cursor.execute(query)

result = cursor.fetchall()

for row in result:
    print(row)
```

```python
#find the actual movie titles

query = """
SELECT DISTINCT
    m.movie_id,
    m.title,
    cr.person_name AS director_name

FROM movies m

JOIN crew cr
    ON m.movie_id = cr.movie_id

WHERE cr.person_name = 'Christopher Nolan'
AND cr.job = 'Director'

ORDER BY m.release_date
"""

cursor.execute(query)

result = cursor.fetchall()

df = pd.DataFrame(
    result,
    columns=["Movie ID", "Movie Title", "Director"]
)

display(df)
```

```python
#Find actors who appeared in these movies
query = """
SELECT DISTINCT
    c.person_id,
    c.actor_name,
    m.title AS movie_title

FROM `cast` c

JOIN crew cr
    ON c.movie_id = cr.movie_id

JOIN movies m
    ON c.movie_id = m.movie_id

WHERE cr.person_name = 'Christopher Nolan'
AND cr.job = 'Director'

ORDER BY m.title, c.actor_name
"""

cursor.execute(query)

result = cursor.fetchall()

df = pd.DataFrame(
    result,
    columns=["Person ID", "Actor Name", "Movie Title"]
)

display(df)
```

```python
#Final step – Get unique actor names directed by Christopher Nolan

query = """
SELECT DISTINCT
    c.person_id,
    c.actor_name

FROM `cast` c

JOIN crew cr
    ON c.movie_id = cr.movie_id

WHERE cr.person_name = 'Christopher Nolan'
AND cr.job = 'Director'

ORDER BY c.actor_name
"""

cursor.execute(query)

result = cursor.fetchall()

df = pd.DataFrame(
    result,
    columns=["Person ID", "Actor Name"]
)

display(df)

print("Total unique actors:", len(df))



# Final results
# Unique actors identified
# 84

# Christopher Nolan movies
# 11

# Actor–movie combinations
# 110

# Unique actors
84
```

## Question 9 – Find the genre with the highest average rating, considering only genres with at least 20 movies.

```python
query = """
SELECT
    g.genre_name,
    AVG(m.vote_average) AS average_rating,
    COUNT(DISTINCT m.movie_id) AS total_movies

FROM movies m

JOIN movie_genres mg
    ON m.movie_id = mg.movie_id

JOIN genres g
    ON mg.genre_id = g.genre_id

WHERE m.vote_average IS NOT NULL

GROUP BY g.genre_id, g.genre_name

HAVING COUNT(DISTINCT m.movie_id) >= 20

ORDER BY average_rating DESC

LIMIT 1
"""

cursor.execute(query)

result = cursor.fetchall()

for row in result:
    print(row)
```

## Question 10 - List the top 10 movies with the highest number of keywords tagged to them. 

```python
query = """
SELECT
    m.title,
    COUNT(DISTINCT mk.keyword_id) AS total_keywords

FROM movies m

JOIN movie_keywords mk
    ON m.movie_id = mk.movie_id

GROUP BY m.movie_id, m.title

ORDER BY total_keywords DESC

LIMIT 10
"""

cursor.execute(query)

result = cursor.fetchall()

# for row in result:
#     print(row)


df = pd.DataFrame(
    result,
    columns=["Movie Title", "Total Keywords"]
)

display(df)
```
