```python
DATA_PATH = "../TMDB_data/"

movies = pd.read_csv(DATA_PATH + "movies.csv")

genres = pd.read_csv(DATA_PATH + "genres.csv")

movie_genres = pd.read_csv(DATA_PATH + "movie_genres.csv")

cast = pd.read_csv(DATA_PATH + "cast.csv")

crew = pd.read_csv(DATA_PATH + "crew.csv")

movie_keywords = pd.read_csv(DATA_PATH + "movie_keywords.csv")
```

```python
datasets = {
    "movies": movies,
    "genres": genres,
    "movie_genres": movie_genres,
    "cast": cast,
    "crew": crew,
    "movie_keywords": movie_keywords
}

for name, df in datasets.items():
    print(f"{name}: {df.shape}")
```
