# Create MySql tables

```python
#Create MySQL tables

#cursor.execute("USE reel_insights")

cursor.execute("""
CREATE TABLE IF NOT EXISTS movies (

    movie_id BIGINT PRIMARY KEY,
    title VARCHAR(255),
    original_language VARCHAR(10),
    release_date DATE,
    runtime INT,
    budget DECIMAL(20,2),
    revenue DECIMAL(20,2),
    popularity DOUBLE,
    vote_average DOUBLE,
    vote_count INT,
    status VARCHAR(50),
    overview TEXT

) ENGINE=InnoDB
""")
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS genres (

    genre_id INT PRIMARY KEY,
    genre_name VARCHAR(100)

) ENGINE=InnoDB
""")
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS movie_genres (

    movie_id BIGINT,
    genre_id INT,

    PRIMARY KEY (movie_id, genre_id),

    FOREIGN KEY (movie_id)
        REFERENCES movies(movie_id),

    FOREIGN KEY (genre_id)
        REFERENCES genres(genre_id)

) ENGINE=InnoDB
""")
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS `cast` (

    cast_row_id BIGINT PRIMARY KEY,
    movie_id BIGINT,
    person_id BIGINT,
    actor_name VARCHAR(255),
    character_name VARCHAR(255),
    cast_order INT,

    FOREIGN KEY (movie_id)
        REFERENCES movies(movie_id)

) ENGINE=InnoDB
""")
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS crew (

    crew_row_id BIGINT PRIMARY KEY,
    movie_id BIGINT,
    person_id BIGINT,
    person_name VARCHAR(255),
    job VARCHAR(100),
    department VARCHAR(100),

    FOREIGN KEY (movie_id)
        REFERENCES movies(movie_id)

) ENGINE=InnoDB
""")
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS movie_keywords (

    movie_keyword_row_id BIGINT AUTO_INCREMENT PRIMARY KEY,
    movie_id BIGINT,
    keyword_id BIGINT,
    keyword_name VARCHAR(255),

    FOREIGN KEY (movie_id)
        REFERENCES movies(movie_id)

) ENGINE=InnoDB
""")
```

# Insert Pandas DataFrames into MySQL

```python
#Dataframe function
def prepare_dataframe(df):

    df = df.copy()

    # Convert datetime to Python date
    if "release_date" in df.columns:

        df["release_date"] = pd.to_datetime(
            df["release_date"],
            errors="coerce"
        ).dt.date

    # Convert missing values into None
    df = df.astype(object).where(
        pd.notna(df),
        None
    )

    return df
```

```python
#To_insert_Data_function

def insert_dataframe(df, table_name):

    df = prepare_dataframe(df)

    columns = list(df.columns)

    column_names = ", ".join(
        f"`{col}`" for col in columns
    )

    placeholders = ", ".join(
        ["%s"] * len(columns)
    )

    query = f"""
    INSERT INTO `{table_name}`
    ({column_names})
    VALUES ({placeholders})
    """

    records = []

    for row in df.itertuples(index=False, name=None):

        clean_row = []

        for value in row:

            if value is None:
                clean_row.append(None)

            elif isinstance(value, (np.integer,)):
                clean_row.append(int(value))

            elif isinstance(value, (np.floating,)):
                clean_row.append(float(value))

            else:
                clean_row.append(value)

        records.append(tuple(clean_row))

    # Insert in batches
    batch_size = 1000

    for i in range(0, len(records), batch_size):

        cursor.executemany(
            query,
            records[i:i + batch_size]
        )

    mydb.commit()

    print(
        f"{table_name}: {len(records)} records inserted"
    )
```

```python
query = """
SELECT
    m.title,
    g.genre_name

FROM movies m

JOIN movie_genres mg
    ON m.movie_id = mg.movie_id

JOIN genres g
    ON mg.genre_id = g.genre_id

LIMIT 10
"""

cursor.execute(query)

result = cursor.fetchall()

for row in result:
    print(row)
```

```python
#see different genres
query = """
SELECT
    m.title,
    g.genre_name
FROM movies m
JOIN movie_genres mg
    ON m.movie_id = mg.movie_id
JOIN genres g
    ON mg.genre_id = g.genre_id
ORDER BY g.genre_name, m.title
LIMIT 20;
"""

cursor.execute(query)

result = cursor.fetchall()

for row in result:
    print(row)
```

```python
#understand genre distribution

query = """
SELECT
    g.genre_name,
    COUNT(*) AS total_movies
FROM movie_genres mg
JOIN genres g
    ON mg.genre_id = g.genre_id
GROUP BY g.genre_name
ORDER BY total_movies DESC;
"""

cursor.execute(query)

result = cursor.fetchall()

for row in result:
    print(row)
```
